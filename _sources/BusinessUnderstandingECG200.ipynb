# Proyek Klasifikasi ECG200

## 1. Business Understanding

### 1.1. Tujuan Proyek (Business Objective)
Tujuan utama dari proyek ini adalah melakukan klasifikasi time series pada data **ECG200** untuk membedakan antara dua kelas kondisi jantung berdasarkan rekaman elektrokardiogram (ECG). Dalam konteks medis, hal ini relevan untuk membantu identifikasi awal kelainan jantung, seperti:
- **Normal** (healthy heart rhythm)
- **Abnormal** (e.g., myocardial infarction or other cardiac anomalies)

### 1.2. Tujuan Analisis (Data Mining Objective)
- Membangun model machine learning yang mampu mengklasifikasikan sinyal ECG ke dalam dua kelas dengan akurasi tinggi.
- Memastikan model dapat digeneralisasi pada data ECG baru yang tidak terlihat sebelumnya.
- Menghasilkan aplikasi berbasis web (Streamlit) yang memungkinkan pengguna mengunggah atau memasukkan data ECG (dalam bentuk array/CSV) dan menerima prediksi kelas secara real-time.

### 1.3. Kriteria Keberhasilan
- Akurasi klasifikasi ≥ 80% (dapat disesuaikan berdasarkan benchmark ECG200).
- Aplikasi Streamlit berfungsi dengan baik dan mudah digunakan.
- Seluruh tahapan CRISP-DM terdokumentasi secara jelas dalam laporan akhir.
