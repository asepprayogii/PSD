# 1. Business Understanding

## 1.1 Latar Belakang

Elektrokardiogram (ECG) adalah rekaman aktivitas kelistrikan jantung yang digunakan secara luas dalam praktik klinis untuk mendeteksi gangguan irama jantung dan kelainan struktural, termasuk **infark miokard** (serangan jantung). Setiap kondisi jantung menghasilkan pola gelombang ECG yang khas—misalnya, elevasi segmen ST atau gelombang Q patologis sering muncul pada pasien dengan kerusakan miokard akut.

Dataset **ECG200** merupakan salah satu *benchmark* standar dalam klasifikasi *time series*, yang awalnya dikembangkan oleh **Robert Olszewski** dalam tesis doktoralnya di **Carnegie Mellon University (2001)**. Dataset ini dirancang untuk menguji kemampuan sistem pengenalan pola dalam membedakan dua kategori denyut jantung berdasarkan morfologi gelombang ECG.

## 1.2 Sumber dan Cara Pengambilan Data

- **Sumber asli**: Rekaman ECG berasal dari pasien dalam lingkungan klinis (rumah sakit/laboratorium medis).  
- **Alat pengambilan**: Sistem perekam ECG medis standar, terdiri dari **elektroda permukaan**, **amplifier**, dan **analog-to-digital converter (ADC)**.  
- **Seleksi data**: Segmen ECG yang merepresentasikan **satu siklus denyut jantung lengkap** dipilih secara manual oleh ahli.  
- **Preprocessing oleh penyedia**:  
  - Dipotong menjadi panjang tetap **96 titik waktu**,  
  - Dinormalisasi (*z-score*: mean ≈ 0, std ≈ 1),  
  - Diberi label: **`1` (normal)** dan **`-1` (abnormal – mengindikasikan infark miokard)**.  
- **Format distribusi**: Disimpan dalam format **ARFF** (*Attribute-Relation File Format*).  

## 1.3 Business Problem

Interpretasi ECG memerlukan pelatihan klinis khusus dan waktu yang tidak sebentar. Di daerah dengan keterbatasan tenaga medis, risiko keterlambatan diagnosis infark miokard meningkat signifikan. Oleh karena itu, diperlukan sistem pendukung keputusan berbasis *machine learning* yang mampu:
- Menganalisis sinyal ECG secara otomatis,
- Memberikan prediksi awal yang andal,
- Diakses melalui antarmuka yang sederhana dan cepat.

## 1.4 Tujuan Analisis

Proyek ini bertujuan untuk:
- Membangun model klasifikasi yang mampu membedakan **sinyal ECG normal** dan **abnormal** menggunakan dataset ECG200,
- Mencapai **F1-score ≥ 0.90** pada data uji sebagai ukuran keseimbangan antara presisi dan *recall*,
- Mengimplementasikan model dalam aplikasi berbasis **Streamlit** yang memungkinkan pengguna mengunggah file CSV berisi 96 nilai ECG dan menerima prediksi *real-time*.

## 1.5 Manfaat Proyek

- **Akademik**: Menjadi studi kasus penerapan CRISP-DM pada data time series medis.  
- **Teknologi**: Menunjukkan bahwa model klasik (seperti Random Forest atau KNN) dapat mencapai performa tinggi pada data kecil jika diproses dengan baik.  
- **Sosial**: Memberikan wawasan tentang potensi AI dalam mendukung deteksi dini penyakit kardiovaskular.  

## 1.6 Ruang Lingkup dan Batasan

- **Dataset**: Hanya menggunakan ECG200 (100 latih, 100 uji, panjang 96, 2 kelas).  
- **Preprocessing**: Normalisasi ulang, penanganan ketidakseimbangan (jika terjadi), tanpa augmentasi.  
- **Modelling**: Perbandingan 6 algoritma klasifikasi.  
- **Deployment**: Aplikasi Streamlit untuk prediksi satu sampel (96 nilai dalam 1 file CSV).  
- **Batasan**:  
  - Bukan alat diagnosis medis,  
  - Tidak menangani noise atau artefak klinis (karena data sudah bersih),  
  - Tidak melakukan deteksi fitur ECG eksplisit (misalnya, QRS complex).
